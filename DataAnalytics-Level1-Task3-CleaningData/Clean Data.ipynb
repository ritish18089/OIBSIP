### Step 5: Generate Data Quality Report

```python

print("=" * 70)
print("                 DATA QUALITY REPORT")
print("=" * 70)

# Dataset Shape
print("\n1. Dataset Shape")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

# Null Values Per Column
print("\n2. Null Values Per Column")
nulls = df.isnull().sum()
null_report = pd.DataFrame({
    "Null Count": nulls,
    "Null %": (nulls / len(df) * 100).round(2)
})
print(null_report[null_report["Null Count"] > 0].sort_values(by="Null Count", ascending=False))

# Duplicate Rows
print("\n3. Duplicate Rows")
duplicates = df.duplicated().sum()
print(f"Total Duplicate Rows: {duplicates}")

# Data Type Issues
print("\n4. Data Types")
dtype_report = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.values
})
print(dtype_report)

# Value Range Anomalies (Numeric Columns)
print("\n5. Value Range Anomalies")
numeric_cols = df.select_dtypes(include=np.number).columns

range_report = pd.DataFrame({
    "Minimum": df[numeric_cols].min(),
    "Maximum": df[numeric_cols].max(),
    "Mean": df[numeric_cols].mean().round(2)
})

print(range_report)

# Detect Potential Outliers Using IQR
print("\nPotential Outliers (IQR Method)")
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {outliers} potential outliers")

print("\n" + "=" * 70)
print("Data Quality Report Generated Successfully")
print("=" * 70)
```

#### Observation:
- The data quality report was generated successfully for the dataset.
- Missing values, duplicate rows, and data types were identified and analyzed.
- Numeric columns were examined for unusual value ranges and potential anomalies.
- These findings provide a clear understanding of the dataset's quality before cleaning.
- The report serves as the foundation for the subsequent data cleaning process.

### Step 6: Missing data handling

```python
# Display missing values before handling
print("Missing Values Before Cleaning:\n")
print(df.isnull().sum())

# 1. Fill numeric columns with Mean
for col in df.select_dtypes(include=['int64', 'float64']).columns:
    df[col] = df[col].fillna(df[col].mean())

# 2. Fill categorical columns with Mode
for col in df.select_dtypes(include=['object']).columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# 3. Forward Fill remaining missing values
df.fillna(method='ffill', inplace=True)

# 4. Delete rows that still contain missing values
df.dropna(inplace=True)

# Display missing values after handling
print("\nMissing Values After Cleaning:\n")
print(df.isnull().sum())
```

#### Observation:
- Mean imputation was used for numerical columns to replace missing values.
- Mode imputation was applied to categorical columns to retain the most frequent category.
- Forward fill was used to replace any remaining missing values.
- Rows with unresolved missing values were removed from the dataset.
- The dataset is now complete and suitable for further analysis.

### Step 7: Duplicate removal

```python
# Duplicate Removal

# Count duplicate rows before removal
duplicate_count = df.duplicated().sum()
print("Duplicate Rows Before Removal:", duplicate_count)

# Remove duplicate rows
df = df.drop_duplicates()

# Count duplicate rows after removal
remaining_duplicates = df.duplicated().sum()
print("Duplicate Rows After Removal:", remaining_duplicates)

# Display the number of rows removed
print("Total Duplicate Rows Removed:", duplicate_count)
```

#### Observation:
- No duplicate rows were found in the dataset before cleaning.
- The drop_duplicates() function was applied to verify and remove any duplicate records.
- Since no duplicates existed, no rows were removed.
- The dataset remained unchanged after the duplicate removal process.
- This confirms that the dataset is free from duplicate entries.

### Step 8: Standardisation: 

```python
# Standardisation

# Remove leading and trailing spaces from text columns
text_cols = df.select_dtypes(include='object').columns
df[text_cols] = df[text_cols].apply(lambda x: x.str.strip())

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Verify
print(df.head())
print(df.dtypes)
```

#### Observation:
- Text columns were standardized by removing leading and trailing spaces.
- The Date column was successfully converted to the datetime format.
- Data formatting is now consistent across the dataset.
- Standardization improves data quality and simplifies further analysis.
- The dataset is ready for the next data cleaning step.

### Step 9: Outlier detection

```python
# Outlier Detection using IQR Method

# Select numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append([col, outliers, "Retained"])

# Display summary
outlier_df = pd.DataFrame(
    outlier_summary,
    columns=["Column", "Outlier Count", "Decision"]
)

print(outlier_df)
```

# Observation:
- Outliers were detected in numeric columns using the IQR method.
- The number of outliers was calculated for each numeric feature.
- Since the dataset contains real sports statistics, the detected outliers were retained.
- Extreme values may represent valid game events rather than data errors.
- Retaining these values preserves the integrity and accuracy of the dataset.

### Step 10: Data type correction

```python
# Data Type Correction

# Convert Date column to datetime
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Convert ID columns to string
id_columns = [col for col in df.columns if 'ID' in col]

for col in id_columns:
    df[col] = df[col].astype('string')

# Convert numeric columns to numeric data types
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Display updated data types
print(df.dtypes)
```

#### Observation:
- The Date column was converted to the datetime format.
- All ID columns were converted to the string data type.
- Numeric columns were verified and converted to appropriate numeric types.
- Invalid values were safely handled using the 'coerce' option where necessary.
- The dataset now has consistent and appropriate data types for analysis.
